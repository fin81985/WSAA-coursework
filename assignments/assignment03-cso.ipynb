```python
import requests
import json

url = "https://ws.cso.ie/public/api.restful/PxStat.Data.Cube_API.ReadDataset/FIQ02/JSON-stat/2.0/en"  # URL to retrieve the FIQ02 dataset from the CSO API in JSON-stat 2.0 format in English

response = requests.get(url)

if response.status_code == 200:# Check if the request to the CSO API was successful
    data = response.json()
    with open("cso.json", "w") as f:
        json.dump(data, f)  # Save the retrieved dataset as json to the file cso.json
    print("Dataset successfully saved to 'cso.json'.")
else:
    print(f"Failed to retrieve dataset. HTTP Status Code: {response.status_code}") # Print the http status code if the request to the CSO API failed

```
